# SGDClassifier

```python
# ============================================================
# CARDIAC MI STAGING — SGD Classifier with Hyperparameter Tuning
# TARGET: Phase (Non-MI, Chronic, Sub-acute, Acute)
# Datasets: 6 folders × {train, val, test}
# Features: Top-10 selected features only
# ============================================================

import subprocess, sys, warnings
warnings.filterwarnings("ignore")

def pip_install(pkg):
    result = subprocess.run([sys.executable, "-m", "pip", "install", "-q", pkg],
                            capture_output=True, text=True)
    print(f"[{'OK' if result.returncode == 0 else 'WARN'}] {pkg}")

pip_install("openpyxl")
pip_install("tabulate")

import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import SGDClassifier
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, classification_report, confusion_matrix
)
from tabulate import tabulate

# ============================================================
# CONFIG
# ============================================================

TARGET_COL = "Phase"

# ── Top-10 features from feature importance chart ──────────────────────────
SELECTED_FEATURES = [
    "Blood_Glucose_mmolL",
    "Diastolic_BP",
    "Chest_Pain",
    "Weighted_Risk_Score",
    "Systolic_BP",
    "Heart_Rate_bpm",
    "Weighted_Symptom_Score",
    "Serum_Creatinine_mgdL",
    "hs_Troponin_I_ngL",
    "Age_Adjusted_Troponin",
]

DATASETS = {
    "Mode Transformation (No SMOTE)": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/mode transformation dataset/train_80_mode_no_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/mode transformation dataset/val_10_mode.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/mode transformation dataset/test_10_mode.xlsx",
    },
    "Mode Transformation + SMOTE": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/mode transformation augmentation dataset/train_80_mode_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/mode transformation augmentation dataset/val_10_mode.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/mode transformation augmentation dataset/test_10_mode.xlsx",
    },
    "Missing Transformation (No SMOTE)": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/missing transformation dataset/train_80_missing_no_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/missing transformation dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/missing transformation dataset/test_10.xlsx",
    },
    "Missing Transformation + SMOTE": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/missing transformation augmentation dataset/train_80_missing_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/missing transformation augmentation dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/missing transformation augmentation dataset/test_10.xlsx",
    },
    "Mode Dataset (No Transformation)": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/mode dataset/train_80_without_transformation.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/mode dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/mode dataset/test_10.xlsx",
    },
    "Missing Dataset": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/missing dataset/train_80_missing.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/missing dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/missing dataset/test_10.xlsx",
    },
}

# ============================================================
# HYPERPARAMETER GRIDS
# SGDClassifier params: loss, penalty, alpha, l1_ratio,
#                       learning_rate, eta0, class_weight, max_iter
# ============================================================

GRID_PARAMS = {
    "loss":          ["hinge", "modified_huber", "log_loss", "squared_hinge"],
    "penalty":       ["l1", "l2", "elasticnet"],
    "alpha":         [0.0001, 0.001, 0.01, 0.1],
    "l1_ratio":      [0.0, 0.25, 0.5, 0.75, 1.0],   # only for elasticnet
    "learning_rate": ["optimal", "adaptive"],
    "class_weight":  [None, "balanced"],
    "max_iter":      [1000],
}

RANDOM_PARAMS = {
    "loss":          ["hinge", "modified_huber", "log_loss",
                      "squared_hinge", "perceptron"],
    "penalty":       ["l1", "l2", "elasticnet"],
    "alpha":         [1e-5, 5e-5, 1e-4, 5e-4, 1e-3, 5e-3, 1e-2, 0.1, 1.0],
    "l1_ratio":      np.linspace(0, 1, 11).tolist(),
    "learning_rate": ["optimal", "adaptive", "invscaling"],
    "eta0":          [0.001, 0.01, 0.1, 1.0],        # used w/ adaptive/invscaling
    "class_weight":  [None, "balanced"],
    "max_iter":      [500, 1000, 2000],
}

N_ITER_RANDOM = 60
CV_FOLDS      = 5
SCORING       = "f1_weighted"
RANDOM_SEED   = 42

# ============================================================
# HELPERS
# ============================================================

def load_dataset(paths, target_col, selected_features):
    """Load train/val/test using only selected features + target."""
    dfs = {}
    for split, path in paths.items():
        df = pd.read_excel(path)
        if target_col not in df.columns:
            for col in df.columns:
                if df[col].dtype == object:
                    df = df.rename(columns={col: target_col})
                    print(f"  [INFO] Renamed '{col}' → '{target_col}' in {split}")
                    break
        dfs[split] = df

    for split, df in dfs.items():
        missing = [f for f in selected_features if f not in df.columns]
        if missing:
            raise ValueError(f"Missing features in {split}: {missing}")

    splits = {}
    for split, df in dfs.items():
        X = df[selected_features]
        y = df[target_col]
        splits[split] = (X, y)

    return splits


def encode_labels(splits):
    """Encode labels only — no imputation or scaling."""
    le = LabelEncoder()
    le.fit(splits["train"][1])

    encoded = {}
    for split, (X, y) in splits.items():
        y_enc = le.transform(y)
        encoded[split] = (X.values, y_enc)

    return encoded, le.classes_.tolist()


def compute_metrics(y_true, y_pred):
    return {
        "Accuracy":      round(accuracy_score(y_true, y_pred), 4),
        "Precision (W)": round(precision_score(y_true, y_pred, average="weighted", zero_division=0), 4),
        "Recall (W)":    round(recall_score(y_true, y_pred,    average="weighted", zero_division=0), 4),
        "F1 Weighted":   round(f1_score(y_true, y_pred,        average="weighted", zero_division=0), 4),
        "F1 Macro":      round(f1_score(y_true, y_pred,        average="macro",    zero_division=0), 4),
    }


def run_search(X_tv, y_tv):
    base = SGDClassifier(random_state=RANDOM_SEED)

    print("    → GridSearchCV ...")
    grid_search = GridSearchCV(
        base, GRID_PARAMS, cv=CV_FOLDS,
        scoring=SCORING, n_jobs=-1, verbose=0, refit=True
    )
    grid_search.fit(X_tv, y_tv)

    print("    → RandomizedSearchCV ...")
    rand_search = RandomizedSearchCV(
        base, RANDOM_PARAMS, n_iter=N_ITER_RANDOM,
        cv=CV_FOLDS, scoring=SCORING, n_jobs=-1,
        verbose=0, refit=True, random_state=RANDOM_SEED
    )
    rand_search.fit(X_tv, y_tv)

    return grid_search, rand_search


# ============================================================
# MAIN LOOP
# ============================================================

all_results = []
all_params  = []

for ds_name, paths in DATASETS.items():
    print("\n" + "█" * 70)
    print(f"  DATASET: {ds_name}")
    print("█" * 70)

    try:
        splits = load_dataset(paths, TARGET_COL, SELECTED_FEATURES)
    except Exception as e:
        print(f"  [ERROR] {e}")
        continue

    print(f"  Using {len(SELECTED_FEATURES)} selected features")
    print(f"  Train: {splits['train'][0].shape[0]}  "
          f"Val: {splits['val'][0].shape[0]}  "
          f"Test: {splits['test'][0].shape[0]}")

    encoded, class_names = encode_labels(splits)
    X_train, y_train = encoded["train"]
    X_val,   y_val   = encoded["val"]
    X_test,  y_test  = encoded["test"]

    # Merge train + val for CV search
    X_tv = np.vstack([X_train, X_val])
    y_tv = np.concatenate([y_train, y_val])
    print(f"  Classes: {class_names}")

    print("  Running hyperparameter search (train + val)...")
    grid_search, rand_search = run_search(X_tv, y_tv)

    print(f"  Grid   best params: {grid_search.best_params_}  CV={grid_search.best_score_:.4f}")
    print(f"  Random best params: {rand_search.best_params_}  CV={rand_search.best_score_:.4f}")

    grid_cv = grid_search.best_score_
    rand_cv = rand_search.best_score_

    for search_name, search in [("GridSearchCV", grid_search), ("RandomizedSearchCV", rand_search)]:
        y_pred   = search.best_estimator_.predict(X_test)
        metrics  = compute_metrics(y_test, y_pred)
        cv_score = search.best_score_
        bp       = search.best_params_

        # ★ = search method with higher CV F1 for this dataset
        winner = "★" if (
            (search_name == "GridSearchCV"       and grid_cv >= rand_cv) or
            (search_name == "RandomizedSearchCV" and rand_cv >  grid_cv)
        ) else ""

        print(f"\n  ── {search_name} TEST RESULTS {winner} ──")
        print(f"     Best params : {bp}")
        print(f"     CV F1       : {cv_score:.4f}")
        print(classification_report(y_test, y_pred, target_names=class_names, zero_division=0))
        print("  Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

        all_results.append({
            "Dataset":     ds_name,
            "Search":      search_name,
            "CV F1":       round(cv_score, 4),
            **metrics
        })

        all_params.append({
            "Dataset":      ds_name,
            "Search":       search_name,
            "loss":         bp.get("loss"),
            "penalty":      bp.get("penalty"),
            "alpha":        bp.get("alpha"),
            "l1_ratio":     bp.get("l1_ratio", "N/A"),
            "learning_rate":bp.get("learning_rate"),
            "eta0":         bp.get("eta0", "N/A"),
            "class_weight": bp.get("class_weight"),
            "max_iter":     bp.get("max_iter"),
            "CV F1":        round(cv_score, 4),
            "_winner":      winner,
        })

# ============================================================
# TABLE 1 — SCORES SUMMARY
# ★ = best value in that metric column across ALL rows
# ============================================================

print("\n\n" + "=" * 115)
print(" TABLE 1 — SCORES SUMMARY  (★ = best in that metric column across all datasets & searches)")
print("=" * 115)

results_df  = pd.DataFrame(all_results)
METRIC_COLS = ["CV F1", "Accuracy", "Precision (W)", "Recall (W)", "F1 Weighted", "F1 Macro"]
best_vals   = {col: results_df[col].max() for col in METRIC_COLS}

score_rows = []
for _, row in results_df.iterrows():
    r = {"Dataset": row["Dataset"], "Search": row["Search"]}
    for col in METRIC_COLS:
        marker = " ★" if row[col] == best_vals[col] else ""
        r[col] = f"{row[col]:.4f}{marker}"
    score_rows.append(r)

print(tabulate(pd.DataFrame(score_rows), headers="keys",
               tablefmt="fancy_grid", showindex=False, numalign="right"))
print("  ★ = best value in that metric column\n")

# ============================================================
# TABLE 2 — BEST PARAMETERS SUMMARY
# ★ = search method with higher CV F1 for that dataset
# ============================================================

print("\n" + "=" * 115)
print(" TABLE 2 — BEST PARAMETERS PER DATASET  (★ = search method with higher CV F1 for that dataset)")
print("=" * 115)

params_df = pd.DataFrame(all_params)

param_rows = []
for _, row in params_df.iterrows():
    param_rows.append({
        "Dataset":       row["Dataset"],
        "Search":        f"{row['Search']} {row['_winner']}".strip(),
        "loss":          row["loss"],
        "penalty":       row["penalty"],
        "alpha":         row["alpha"],
        "l1_ratio":      row["l1_ratio"],
        "learning_rate": row["learning_rate"],
        "eta0":          row["eta0"],
        "class_weight":  row["class_weight"],
        "max_iter":      row["max_iter"],
        "CV F1":         f"{row['CV F1']:.4f}",
    })

print(tabulate(pd.DataFrame(param_rows), headers="keys",
               tablefmt="fancy_grid", showindex=False, numalign="right"))
print("  ★ = search method that achieved higher CV F1 for that dataset\n")

# ============================================================
# PER-METRIC WINNER SUMMARY
# ============================================================

print("\n" + "─" * 80)
print(" BEST CONFIGURATION PER METRIC")
print("─" * 80)
for col in METRIC_COLS:
    best_row = results_df.loc[results_df[col].idxmax()]
    print(f"  {col:<18}  {best_row[col]:.4f}  →  {best_row['Dataset']}  [{best_row['Search']}]")

# Save
results_df.to_csv("/kaggle/working/sgd_scores.csv", index=False)
params_df.drop(columns=["_winner"]).to_csv("/kaggle/working/sgd_best_params.csv", index=False)
print("\nSaved: /kaggle/working/sgd_scores.csv  &  /kaggle/working/sgd_best_params.csv")
print("\n✅  All done!")
```

```python
# ============================================================
# CARDIAC MI STAGING — Decision Tree with Hyperparameter Tuning
# TARGET: Phase (Non-MI, Chronic, Sub-acute, Acute)
# Datasets: 6 folders × {train, val, test}
# Features: Top-10 selected features only
# ============================================================

import subprocess, sys, warnings
warnings.filterwarnings("ignore")

def pip_install(pkg):
    result = subprocess.run([sys.executable, "-m", "pip", "install", "-q", pkg],
                            capture_output=True, text=True)
    print(f"[{'OK' if result.returncode == 0 else 'WARN'}] {pkg}")

pip_install("openpyxl")
pip_install("tabulate")

import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, classification_report, confusion_matrix
)
from tabulate import tabulate

# ============================================================
# CONFIG
# ============================================================

TARGET_COL = "Phase"

# ── Top-10 features from feature importance chart ──────────────────────────
SELECTED_FEATURES = [
    "Blood_Glucose_mmolL",
    "Diastolic_BP",
    "Chest_Pain",
    "Weighted_Risk_Score",
    "Systolic_BP",
    "Heart_Rate_bpm",
    "Weighted_Symptom_Score",
    "Serum_Creatinine_mgdL",
    "hs_Troponin_I_ngL",
    "Age_Adjusted_Troponin",
]

DATASETS = {
    "Mode Transformation (No SMOTE)": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/mode transformation dataset/train_80_mode_no_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/mode transformation dataset/val_10_mode.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/mode transformation dataset/test_10_mode.xlsx",
    },
    "Mode Transformation + SMOTE": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/mode transformation augmentation dataset/train_80_mode_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/mode transformation augmentation dataset/val_10_mode.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/mode transformation augmentation dataset/test_10_mode.xlsx",
    },
    "Missing Transformation (No SMOTE)": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/missing transformation dataset/train_80_missing_no_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/missing transformation dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/missing transformation dataset/test_10.xlsx",
    },
    "Missing Transformation + SMOTE": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/missing transformation augmentation dataset/train_80_missing_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/missing transformation augmentation dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/missing transformation augmentation dataset/test_10.xlsx",
    },
    "Mode Dataset (No Transformation)": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/mode dataset/train_80_without_transformation.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/mode dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/mode dataset/test_10.xlsx",
    },
    "Missing Dataset": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/missing dataset/train_80_missing.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/missing dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/missing dataset/test_10.xlsx",
    },
}

# ============================================================
# HYPERPARAMETER GRIDS
# DecisionTreeClassifier params: criterion, splitter, max_depth,
#   min_samples_split, min_samples_leaf, max_features, class_weight
# ============================================================

GRID_PARAMS = {
    "criterion":          ["gini", "entropy", "log_loss"],
    "splitter":           ["best", "random"],
    "max_depth":          [None, 3, 5, 10, 20],
    "min_samples_split":  [2, 5, 10],
    "min_samples_leaf":   [1, 2, 4],
    "max_features":       [None, "sqrt", "log2"],
    "class_weight":       [None, "balanced"],
}

RANDOM_PARAMS = {
    "criterion":          ["gini", "entropy", "log_loss"],
    "splitter":           ["best", "random"],
    "max_depth":          [None, 2, 3, 5, 7, 10, 15, 20, 30],
    "min_samples_split":  [2, 3, 5, 8, 10, 15, 20],
    "min_samples_leaf":   [1, 2, 3, 4, 5, 8, 10],
    "max_features":       [None, "sqrt", "log2", 0.5, 0.7],
    "class_weight":       [None, "balanced"],
    "ccp_alpha":          [0.0, 0.001, 0.005, 0.01, 0.05],  # pruning
}

N_ITER_RANDOM = 60
CV_FOLDS      = 5
SCORING       = "f1_weighted"
RANDOM_SEED   = 42

# ============================================================
# HELPERS
# ============================================================

def load_dataset(paths, target_col, selected_features):
    dfs = {}
    for split, path in paths.items():
        df = pd.read_excel(path)
        if target_col not in df.columns:
            for col in df.columns:
                if df[col].dtype == object:
                    df = df.rename(columns={col: target_col})
                    print(f"  [INFO] Renamed '{col}' → '{target_col}' in {split}")
                    break
        dfs[split] = df

    for split, df in dfs.items():
        missing = [f for f in selected_features if f not in df.columns]
        if missing:
            raise ValueError(f"Missing features in {split}: {missing}")

    splits = {}
    for split, df in dfs.items():
        X = df[selected_features]
        y = df[target_col]
        splits[split] = (X, y)

    return splits


def encode_labels(splits):
    le = LabelEncoder()
    le.fit(splits["train"][1])

    encoded = {}
    for split, (X, y) in splits.items():
        y_enc = le.transform(y)
        encoded[split] = (X.values, y_enc)

    return encoded, le.classes_.tolist()


def compute_metrics(y_true, y_pred):
    return {
        "Accuracy":      round(accuracy_score(y_true, y_pred), 4),
        "Precision (W)": round(precision_score(y_true, y_pred, average="weighted", zero_division=0), 4),
        "Recall (W)":    round(recall_score(y_true, y_pred,    average="weighted", zero_division=0), 4),
        "F1 Weighted":   round(f1_score(y_true, y_pred,        average="weighted", zero_division=0), 4),
        "F1 Macro":      round(f1_score(y_true, y_pred,        average="macro",    zero_division=0), 4),
    }


def run_search(X_tv, y_tv):
    base = DecisionTreeClassifier(random_state=RANDOM_SEED)

    print("    → GridSearchCV ...")
    grid_search = GridSearchCV(
        base, GRID_PARAMS, cv=CV_FOLDS,
        scoring=SCORING, n_jobs=-1, verbose=0, refit=True
    )
    grid_search.fit(X_tv, y_tv)

    print("    → RandomizedSearchCV ...")
    rand_search = RandomizedSearchCV(
        base, RANDOM_PARAMS, n_iter=N_ITER_RANDOM,
        cv=CV_FOLDS, scoring=SCORING, n_jobs=-1,
        verbose=0, refit=True, random_state=RANDOM_SEED
    )
    rand_search.fit(X_tv, y_tv)

    return grid_search, rand_search


# ============================================================
# MAIN LOOP
# ============================================================

all_results = []
all_params  = []

for ds_name, paths in DATASETS.items():
    print("\n" + "█" * 70)
    print(f"  DATASET: {ds_name}")
    print("█" * 70)

    try:
        splits = load_dataset(paths, TARGET_COL, SELECTED_FEATURES)
    except Exception as e:
        print(f"  [ERROR] {e}")
        continue

    print(f"  Using {len(SELECTED_FEATURES)} selected features")
    print(f"  Train: {splits['train'][0].shape[0]}  "
          f"Val: {splits['val'][0].shape[0]}  "
          f"Test: {splits['test'][0].shape[0]}")

    encoded, class_names = encode_labels(splits)
    X_train, y_train = encoded["train"]
    X_val,   y_val   = encoded["val"]
    X_test,  y_test  = encoded["test"]

    X_tv = np.vstack([X_train, X_val])
    y_tv = np.concatenate([y_train, y_val])
    print(f"  Classes: {class_names}")

    print("  Running hyperparameter search (train + val)...")
    grid_search, rand_search = run_search(X_tv, y_tv)

    print(f"  Grid   best params: {grid_search.best_params_}  CV={grid_search.best_score_:.4f}")
    print(f"  Random best params: {rand_search.best_params_}  CV={rand_search.best_score_:.4f}")

    grid_cv = grid_search.best_score_
    rand_cv = rand_search.best_score_

    for search_name, search in [("GridSearchCV", grid_search), ("RandomizedSearchCV", rand_search)]:
        y_pred   = search.best_estimator_.predict(X_test)
        metrics  = compute_metrics(y_test, y_pred)
        cv_score = search.best_score_
        bp       = search.best_params_

        winner = "★" if (
            (search_name == "GridSearchCV"       and grid_cv >= rand_cv) or
            (search_name == "RandomizedSearchCV" and rand_cv >  grid_cv)
        ) else ""

        print(f"\n  ── {search_name} TEST RESULTS {winner} ──")
        print(f"     Best params : {bp}")
        print(f"     CV F1       : {cv_score:.4f}")
        print(classification_report(y_test, y_pred, target_names=class_names, zero_division=0))
        print("  Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

        all_results.append({
            "Dataset":  ds_name,
            "Search":   search_name,
            "CV F1":    round(cv_score, 4),
            **metrics
        })

        all_params.append({
            "Dataset":           ds_name,
            "Search":            search_name,
            "criterion":         bp.get("criterion"),
            "splitter":          bp.get("splitter"),
            "max_depth":         bp.get("max_depth", "None"),
            "min_samples_split": bp.get("min_samples_split"),
            "min_samples_leaf":  bp.get("min_samples_leaf"),
            "max_features":      bp.get("max_features", "None"),
            "class_weight":      bp.get("class_weight"),
            "ccp_alpha":         bp.get("ccp_alpha", "N/A"),
            "CV F1":             round(cv_score, 4),
            "_winner":           winner,
        })

# ============================================================
# TABLE 1 — SCORES SUMMARY
# ★ = best value in that metric column across ALL rows
# ============================================================

print("\n\n" + "=" * 115)
print(" TABLE 1 — SCORES SUMMARY  (★ = best in that metric column across all datasets & searches)")
print("=" * 115)

results_df  = pd.DataFrame(all_results)
METRIC_COLS = ["CV F1", "Accuracy", "Precision (W)", "Recall (W)", "F1 Weighted", "F1 Macro"]
best_vals   = {col: results_df[col].max() for col in METRIC_COLS}

score_rows = []
for _, row in results_df.iterrows():
    r = {"Dataset": row["Dataset"], "Search": row["Search"]}
    for col in METRIC_COLS:
        marker = " ★" if row[col] == best_vals[col] else ""
        r[col] = f"{row[col]:.4f}{marker}"
    score_rows.append(r)

print(tabulate(pd.DataFrame(score_rows), headers="keys",
               tablefmt="fancy_grid", showindex=False, numalign="right"))
print("  ★ = best value in that metric column\n")

# ============================================================
# TABLE 2 — BEST PARAMETERS SUMMARY
# ★ = search method with higher CV F1 for that dataset
# ============================================================

print("\n" + "=" * 115)
print(" TABLE 2 — BEST PARAMETERS PER DATASET  (★ = search method with higher CV F1 for that dataset)")
print("=" * 115)

params_df = pd.DataFrame(all_params)

param_rows = []
for _, row in params_df.iterrows():
    param_rows.append({
        "Dataset":           row["Dataset"],
        "Search":            f"{row['Search']} {row['_winner']}".strip(),
        "criterion":         row["criterion"],
        "splitter":          row["splitter"],
        "max_depth":         row["max_depth"],
        "min_samples_split": row["min_samples_split"],
        "min_samples_leaf":  row["min_samples_leaf"],
        "max_features":      row["max_features"],
        "class_weight":      row["class_weight"],
        "ccp_alpha":         row["ccp_alpha"],
        "CV F1":             f"{row['CV F1']:.4f}",
    })

print(tabulate(pd.DataFrame(param_rows), headers="keys",
               tablefmt="fancy_grid", showindex=False, numalign="right"))
print("  ★ = search method that achieved higher CV F1 for that dataset\n")

# ============================================================
# PER-METRIC WINNER SUMMARY
# ============================================================

print("\n" + "─" * 80)
print(" BEST CONFIGURATION PER METRIC")
print("─" * 80)
for col in METRIC_COLS:
    best_row = results_df.loc[results_df[col].idxmax()]
    print(f"  {col:<18}  {best_row[col]:.4f}  →  {best_row['Dataset']}  [{best_row['Search']}]")

# Save
results_df.to_csv("/kaggle/working/dt_scores.csv", index=False)
params_df.drop(columns=["_winner"]).to_csv("/kaggle/working/dt_best_params.csv", index=False)
print("\nSaved: /kaggle/working/dt_scores.csv  &  /kaggle/working/dt_best_params.csv")
print("\n✅  All done!")
```

```python
# ============================================================
# CARDIAC MI STAGING — Random Forest with Hyperparameter Tuning
# TARGET: Phase (Non-MI, Chronic, Sub-acute, Acute)
# Datasets: 6 folders × {train, val, test}
# Features: Top-10 selected features only
# ============================================================

import subprocess, sys, warnings
warnings.filterwarnings("ignore")

def pip_install(pkg):
    result = subprocess.run([sys.executable, "-m", "pip", "install", "-q", pkg],
                            capture_output=True, text=True)
    print(f"[{'OK' if result.returncode == 0 else 'WARN'}] {pkg}")

pip_install("openpyxl")
pip_install("tabulate")

import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, classification_report, confusion_matrix
)
from tabulate import tabulate

# ============================================================
# CONFIG
# ============================================================

TARGET_COL = "Phase"

# ── Top-10 features from feature importance chart ──────────────────────────
SELECTED_FEATURES = [
    "Blood_Glucose_mmolL",
    "Diastolic_BP",
    "Chest_Pain",
    "Weighted_Risk_Score",
    "Systolic_BP",
    "Heart_Rate_bpm",
    "Weighted_Symptom_Score",
    "Serum_Creatinine_mgdL",
    "hs_Troponin_I_ngL",
    "Age_Adjusted_Troponin",
]

DATASETS = {
    "Mode Transformation (No SMOTE)": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/mode transformation dataset/train_80_mode_no_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/mode transformation dataset/val_10_mode.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/mode transformation dataset/test_10_mode.xlsx",
    },
    "Mode Transformation + SMOTE": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/mode transformation augmentation dataset/train_80_mode_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/mode transformation augmentation dataset/val_10_mode.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/mode transformation augmentation dataset/test_10_mode.xlsx",
    },
    "Missing Transformation (No SMOTE)": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/missing transformation dataset/train_80_missing_no_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/missing transformation dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/missing transformation dataset/test_10.xlsx",
    },
    "Missing Transformation + SMOTE": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/missing transformation augmentation dataset/train_80_missing_smote.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/missing transformation augmentation dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/missing transformation augmentation dataset/test_10.xlsx",
    },
    "Mode Dataset (No Transformation)": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/mode dataset/train_80_without_transformation.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/mode dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/mode dataset/test_10.xlsx",
    },
    "Missing Dataset": {
        "train": "/kaggle/input/datasets/fairoooz/sagorer/missing dataset/train_80_missing.xlsx",
        "val":   "/kaggle/input/datasets/fairoooz/sagorer/missing dataset/val_10.xlsx",
        "test":  "/kaggle/input/datasets/fairoooz/sagorer/missing dataset/test_10.xlsx",
    },
}

# ============================================================
# HYPERPARAMETER GRIDS
# RandomForestClassifier params: n_estimators, criterion,
#   max_depth, min_samples_split, min_samples_leaf,
#   max_features, bootstrap, class_weight
# ============================================================

GRID_PARAMS = {
    "n_estimators":      [50, 100, 200],
    "criterion":         ["gini", "entropy"],
    "max_depth":         [None, 5, 10, 20],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf":  [1, 2, 4],
    "max_features":      ["sqrt", "log2"],
    "bootstrap":         [True, False],
    "class_weight":      [None, "balanced"],
}

RANDOM_PARAMS = {
    "n_estimators":      [50, 100, 200, 300, 500],
    "criterion":         ["gini", "entropy", "log_loss"],
    "max_depth":         [None, 3, 5, 7, 10, 15, 20, 30],
    "min_samples_split": [2, 3, 5, 8, 10, 15, 20],
    "min_samples_leaf":  [1, 2, 3, 4, 5, 8],
    "max_features":      ["sqrt", "log2", None, 0.3, 0.5, 0.7],
    "bootstrap":         [True, False],
    "class_weight":      [None, "balanced", "balanced_subsample"],
    "ccp_alpha":         [0.0, 0.001, 0.005, 0.01],
}

N_ITER_RANDOM = 60
CV_FOLDS      = 5
SCORING       = "f1_weighted"
RANDOM_SEED   = 42

# ============================================================
# HELPERS
# ============================================================

def load_dataset(paths, target_col, selected_features):
    dfs = {}
    for split, path in paths.items():
        df = pd.read_excel(path)
        if target_col not in df.columns:
            for col in df.columns:
                if df[col].dtype == object:
                    df = df.rename(columns={col: target_col})
                    print(f"  [INFO] Renamed '{col}' → '{target_col}' in {split}")
                    break
        dfs[split] = df

    for split, df in dfs.items():
        missing = [f for f in selected_features if f not in df.columns]
        if missing:
            raise ValueError(f"Missing features in {split}: {missing}")

    splits = {}
    for split, df in dfs.items():
        X = df[selected_features]
        y = df[target_col]
        splits[split] = (X, y)

    return splits


def encode_labels(splits):
    le = LabelEncoder()
    le.fit(splits["train"][1])

    encoded = {}
    for split, (X, y) in splits.items():
        y_enc = le.transform(y)
        encoded[split] = (X.values, y_enc)

    return encoded, le.classes_.tolist()


def compute_metrics(y_true, y_pred):
    return {
        "Accuracy":      round(accuracy_score(y_true, y_pred), 4),
        "Precision (W)": round(precision_score(y_true, y_pred, average="weighted", zero_division=0), 4),
        "Recall (W)":    round(recall_score(y_true, y_pred,    average="weighted", zero_division=0), 4),
        "F1 Weighted":   round(f1_score(y_true, y_pred,        average="weighted", zero_division=0), 4),
        "F1 Macro":      round(f1_score(y_true, y_pred,        average="macro",    zero_division=0), 4),
    }


def run_search(X_tv, y_tv):
    base = RandomForestClassifier(random_state=RANDOM_SEED, n_jobs=-1)

    print("    → GridSearchCV ...")
    grid_search = GridSearchCV(
        base, GRID_PARAMS, cv=CV_FOLDS,
        scoring=SCORING, n_jobs=-1, verbose=0, refit=True
    )
    grid_search.fit(X_tv, y_tv)

    print("    → RandomizedSearchCV ...")
    rand_search = RandomizedSearchCV(
        base, RANDOM_PARAMS, n_iter=N_ITER_RANDOM,
        cv=CV_FOLDS, scoring=SCORING, n_jobs=-1,
        verbose=0, refit=True, random_state=RANDOM_SEED
    )
    rand_search.fit(X_tv, y_tv)

    return grid_search, rand_search


# ============================================================
# MAIN LOOP
# ============================================================

all_results = []
all_params  = []

for ds_name, paths in DATASETS.items():
    print("\n" + "█" * 70)
    print(f"  DATASET: {ds_name}")
    print("█" * 70)

    try:
        splits = load_dataset(paths, TARGET_COL, SELECTED_FEATURES)
    except Exception as e:
        print(f"  [ERROR] {e}")
        continue

    print(f"  Using {len(SELECTED_FEATURES)} selected features")
    print(f"  Train: {splits['train'][0].shape[0]}  "
          f"Val: {splits['val'][0].shape[0]}  "
          f"Test: {splits['test'][0].shape[0]}")

    encoded, class_names = encode_labels(splits)
    X_train, y_train = encoded["train"]
    X_val,   y_val   = encoded["val"]
    X_test,  y_test  = encoded["test"]

    X_tv = np.vstack([X_train, X_val])
    y_tv = np.concatenate([y_train, y_val])
    print(f"  Classes: {class_names}")

    print("  Running hyperparameter search (train + val)...")
    grid_search, rand_search = run_search(X_tv, y_tv)

    print(f"  Grid   best params: {grid_search.best_params_}  CV={grid_search.best_score_:.4f}")
    print(f"  Random best params: {rand_search.best_params_}  CV={rand_search.best_score_:.4f}")

    grid_cv = grid_search.best_score_
    rand_cv = rand_search.best_score_

    for search_name, search in [("GridSearchCV", grid_search), ("RandomizedSearchCV", rand_search)]:
        y_pred   = search.best_estimator_.predict(X_test)
        metrics  = compute_metrics(y_test, y_pred)
        cv_score = search.best_score_
        bp       = search.best_params_

        winner = "★" if (
            (search_name == "GridSearchCV"       and grid_cv >= rand_cv) or
            (search_name == "RandomizedSearchCV" and rand_cv >  grid_cv)
        ) else ""

        print(f"\n  ── {search_name} TEST RESULTS {winner} ──")
        print(f"     Best params : {bp}")
        print(f"     CV F1       : {cv_score:.4f}")
        print(classification_report(y_test, y_pred, target_names=class_names, zero_division=0))
        print("  Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

        all_results.append({
            "Dataset":  ds_name,
            "Search":   search_name,
            "CV F1":    round(cv_score, 4),
            **metrics
        })

        all_params.append({
            "Dataset":           ds_name,
            "Search":            search_name,
            "n_estimators":      bp.get("n_estimators"),
            "criterion":         bp.get("criterion"),
            "max_depth":         bp.get("max_depth", "None"),
            "min_samples_split": bp.get("min_samples_split"),
            "min_samples_leaf":  bp.get("min_samples_leaf"),
            "max_features":      bp.get("max_features"),
            "bootstrap":         bp.get("bootstrap"),
            "class_weight":      bp.get("class_weight"),
            "ccp_alpha":         bp.get("ccp_alpha", "N/A"),
            "CV F1":             round(cv_score, 4),
            "_winner":           winner,
        })

# ============================================================
# TABLE 1 — SCORES SUMMARY
# ★ = best value in that metric column across ALL rows
# ============================================================

print("\n\n" + "=" * 115)
print(" TABLE 1 — SCORES SUMMARY  (★ = best in that metric column across all datasets & searches)")
print("=" * 115)

results_df  = pd.DataFrame(all_results)
METRIC_COLS = ["CV F1", "Accuracy", "Precision (W)", "Recall (W)", "F1 Weighted", "F1 Macro"]
best_vals   = {col: results_df[col].max() for col in METRIC_COLS}

score_rows = []
for _, row in results_df.iterrows():
    r = {"Dataset": row["Dataset"], "Search": row["Search"]}
    for col in METRIC_COLS:
        marker = " ★" if row[col] == best_vals[col] else ""
        r[col] = f"{row[col]:.4f}{marker}"
    score_rows.append(r)

print(tabulate(pd.DataFrame(score_rows), headers="keys",
               tablefmt="fancy_grid", showindex=False, numalign="right"))
print("  ★ = best value in that metric column\n")

# ============================================================
# TABLE 2 — BEST PARAMETERS SUMMARY
# ★ = search method with higher CV F1 for that dataset
# ============================================================

print("\n" + "=" * 115)
print(" TABLE 2 — BEST PARAMETERS PER DATASET  (★ = search method with higher CV F1 for that dataset)")
print("=" * 115)

params_df = pd.DataFrame(all_params)

param_rows = []
for _, row in params_df.iterrows():
    param_rows.append({
        "Dataset":           row["Dataset"],
        "Search":            f"{row['Search']} {row['_winner']}".strip(),
        "n_estimators":      row["n_estimators"],
        "criterion":         row["criterion"],
        "max_depth":         row["max_depth"],
        "min_samples_split": row["min_samples_split"],
        "min_samples_leaf":  row["min_samples_leaf"],
        "max_features":      row["max_features"],
        "bootstrap":         row["bootstrap"],
        "class_weight":      row["class_weight"],
        "ccp_alpha":         row["ccp_alpha"],
        "CV F1":             f"{row['CV F1']:.4f}",
    })

print(tabulate(pd.DataFrame(param_rows), headers="keys",
               tablefmt="fancy_grid", showindex=False, numalign="right"))
print("  ★ = search method that achieved higher CV F1 for that dataset\n")

# ============================================================
# PER-METRIC WINNER SUMMARY
# ============================================================

print("\n" + "─" * 80)
print(" BEST CONFIGURATION PER METRIC")
print("─" * 80)
for col in METRIC_COLS:
    best_row = results_df.loc[results_df[col].idxmax()]
    print(f"  {col:<18}  {best_row[col]:.4f}  →  {best_row['Dataset']}  [{best_row['Search']}]")

# Save
results_df.to_csv("/kaggle/working/rf_scores.csv", index=False)
params_df.drop(columns=["_winner"]).to_csv("/kaggle/working/rf_best_params.csv", index=False)
print("\nSaved: /kaggle/working/rf_scores.csv  &  /kaggle/working/rf_best_params.csv")
print("\n✅  All done!")
```
